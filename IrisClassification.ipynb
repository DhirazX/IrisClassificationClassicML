# Imports

```python
#Basic Imports
import os

#Extra Imports
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
#Import dataset

data = datasets.load_iris()
data.keys()
```

# Convert to dataframe


```python
df = pd.DataFrame(data['data'], columns=data['feature_names'])
df['target'] = data['target']
df.head()
```

```python
df.describe()
```

# Distribution of features and target

```python
col = "sepal length (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "sepal width (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "petal length (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

```python
col = "petal width (cm)"
df[col].hist()
plt.suptitle(col)
plt.show()
```

# Relationship of features with target

```python
# Adding another column for target names
df['target_name'] = df['target'].map({0: "setosa", 1: "versicolor", 2: "virginica"})
df
```

```python
col = "sepal length (cm)"
sns.relplot(x = col, y="target", data=df, hue="target_name")
_ = plt.suptitle(col)
```

```python
col = "sepal width (cm)"
sns.relplot(x = col, y="target", data=df, hue="target_name")
_ = plt.suptitle(col)
```

```python
col = "petal length (cm)"
sns.relplot(x = col, y="target", data=df, hue="target_name")
_ = plt.suptitle(col)
```

```python
col = "petal width (cm)"
sns.relplot(x = col, y="target", data=df, hue="target_name")
_ = plt.suptitle(col)
```
